```python
from scipy.stats import chi2_contingency
import pandas as pd
import numpy as np
```

```python
boxoffice = pd.read_csv('2026-04-30_boxoffice.csv')
```

```python
tmdb = pd.read_csv('tmdb_data.csv')
```

```python
df= pd.merge(tmdb, boxoffice, on='title')
print(df.head())
```

```python
df['total_gross'] = (
    df['total_gross']
    .replace('[\$,]', '', regex=True)  # remove $ and commas
    .astype(float)
)
```

```python
bins = [0,50_000_000,250_000_000, 450_000_000, df['total_gross'].max()]
labels= ['Low', 'Medium', 'High','Very High']
df['revenue_category']= pd.cut(df['total_gross'], bins=bins, labels=labels)
```

```python
df['rating_category'] = pd.qcut(
    df['rating'],
    q=3,
    labels=['Low', 'Medium', 'High']
)
```

```python
contingency_table = pd.crosstab(
    df['revenue_category'],
    df['rating_category']
)

print(contingency_table)
```

```python
chi2, p, dof, expected = chi2_contingency(contingency_table)
print('Chi-square statistic:', chi2)
print('p-value:', p)
print('Degrees of freedom:', dof)
```

```python
alpha = 0.05
if p < alpha:
    print('Significant association between box office and nominations')
else:
    print('No significant association')
```
